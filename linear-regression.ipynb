```python
from sklearn import linear_model
```

```python
df=pd.read_csv("/kaggle/input/datasets/kazi23/homeprices/homeprices.csv - Sheet1.csv")
df
```

```python
import math

medium_bedrooms=math.floor(df['bedrooms'].median())
print(medium_bedrooms)
```

```python
df['bedrooms']=df['bedrooms'].fillna(medium_bedrooms)
df
```

```python
reg=linear_model.LinearRegression() # object of the model
reg.fit(df[['area','bedrooms','age']],df['price']) # train the model using the dataset
```

```python
reg.coef_ # the weight to each variable
```

```python
reg.intercept_
```

```python
reg.predict([[3000,3,40]])
```
